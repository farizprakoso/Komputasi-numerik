```python
# Identitas Mahasiswa: Al Fariz Prakoso (NIM: 202511420018)
# Mengimpor library NumPy untuk mempermudah operasi matriks dan array numerik
import numpy as np

# Mendefinisikan matriks koefisien (3x3) dari sistem persamaan:
# 2x + 3y -  z = 4
# 3x -  y + 2z = 11
#  x + 2y + 3z = 13
matriks_koef = np.array([
    [2.0,  3.0, -1.0],
    [3.0, -1.0,  2.0],
    [1.0,  2.0,  3.0]
], dtype=float)

vektor_konstanta = np.array([4.0, 11.0, 13.0], dtype=float)

matriks_teraugmentasi = np.hstack([matriks_koef, vektor_konstanta.reshape(-1, 1)])

print("=== MATRIKS TERAUGMENTASI AWAL ===")
print(matriks_teraugmentasi)
```

```python
jumlah_persamaan = len(vektor_konstanta)

for i in range(jumlah_persamaan):
    if matriks_teraugmentasi[i][i] == 0.0:
        raise ValueError("Pivot bernilai nol ditemukan, proses eliminasi tidak dapat dilanjutkan.")

    for j in range(i + 1, jumlah_persamaan):
        faktor_pengali = matriks_teraugmentasi[j][i] / matriks_teraugmentasi[i][i]
        matriks_teraugmentasi[j] = matriks_teraugmentasi[j] - faktor_pengali * matriks_teraugmentasi[i]

print("=== MATRIKS SEGITIGA ATAS (HASIL ELIMINASI MAJU) ===")
print(np.round(matriks_teraugmentasi, 4))
```

```python
hasil_variabel = np.zeros(jumlah_persamaan)

for i in range(jumlah_persamaan - 1, -1, -1):
    jumlah_suku_lain = sum(matriks_teraugmentasi[i][j] * hasil_variabel[j] for j in range(i + 1, jumlah_persamaan))
    hasil_variabel[i] = (matriks_teraugmentasi[i][jumlah_persamaan] - jumlah_suku_lain) / matriks_teraugmentasi[i][i]

print("=== SOLUSI AKHIR SPL3V ===")
print(f"Nilai x = {hasil_variabel[0]:.4f}")
print(f"Nilai y = {hasil_variabel[1]:.4f}")
print(f"Nilai z = {hasil_variabel[2]:.4f}")
```
