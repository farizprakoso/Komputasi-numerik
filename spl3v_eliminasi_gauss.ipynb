```python
# TUGAS KOMPUTASI NUMERIK - SPL3V (ELIMINASI GAUSS)
# Nama : Al Fariz Prakoso | NIM : 202511420018

import numpy as np

# Definisi Matriks Koefisien (A) dan Matriks Konstanta (B)
A = np.array([
    [2.0, 1.0, -1.0],
    [-3.0, -1.0, 2.0],
    [-2.0, 1.0, 2.0]
], dtype=float)

B = np.array([8.0, -11.0, -3.0], dtype=float)

# Matriks Teraugmentasi [A|B]
aug = np.hstack([A, B.reshape(-1, 1)])
print("Matriks Teraugmentasi Awal:\n", aug)
```

```python
# Proses Eliminasi Maju menjadi Matriks Segitiga Atas
n = len(B)

for i in range(n):
    if aug[i][i] == 0.0:
        raise ValueError("Pivot bernilai nol.")

    for j in range(i + 1, n):
        faktor = aug[j][i] / aug[i][i]
        aug[j] = aug[j] - faktor * aug[i]

print("Matriks Segitiga Atas:\n", np.round(aug, 4))
```

```python
# Proses Substitusi Balik
x_sol = np.zeros(n)

for i in range(n - 1, -1, -1):
    suku_diketahui = sum(aug[i][j] * x_sol[j] for j in range(i + 1, n))
    x_sol[i] = (aug[i][n] - suku_diketahui) / aug[i][i]

print("=== SOLUSI SPL3V ===")
print(f"Nilai x = {x_sol[0]:.4f}")
print(f"Nilai y = {x_sol[1]:.4f}")
print(f"Nilai z = {x_sol[2]:.4f}")
```
