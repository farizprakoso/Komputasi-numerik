```python
# Identitas Mahasiswa: Al Fariz Prakoso (NIM: 202511420018)
# Mengimpor NumPy untuk perhitungan numerik dan Matplotlib untuk menggambar grafik fungsi
import numpy as np
import matplotlib.pyplot as plt

def persamaan_non_linear(x):
    return x**3 - 2*x - 5

sumbu_x = np.linspace(1.0, 3.5, 150)
nilai_y = persamaan_non_linear(sumbu_x)

plt.figure(figsize=(8, 4))
plt.plot(sumbu_x, nilai_y, label='f(x) = x^3 - 2x - 5', color='navy')
plt.axhline(0, color='crimson', linestyle='--', label='Garis Sumbu y = 0')
plt.title("Visualisasi Kurva Persamaan Non-Linear (Metode Bisection)")
plt.xlabel("Nilai x")
plt.ylabel("Nilai f(x)")
plt.grid(True)
plt.legend()
# Menampilkan grafik ke layar output Colab
plt.show()
```

```python
# Mendefinisikan fungsi metode_bagi_dua dengan parameter batas bawah, batas atas, dan batas galat toleransi
def metode_bagi_dua(batas_bawah, batas_atas, batas_galat=1e-4):
    if persamaan_non_linear(batas_bawah) * persamaan_non_linear(batas_atas) >= 0:
        print("Interval awal tidak valid karena f(a) * f(b) >= 0.")
        return None

    iterasi_ke = 1

    while (batas_atas - batas_bawah) / 2.0 > batas_galat:
        titik_tengah_c = (batas_bawah + batas_atas) / 2.0
        nilai_fc = persamaan_non_linear(titik_tengah_c)

        print(f"Iterasi {iterasi_ke:2d} | a = {batas_bawah:.5f} | b = {batas_atas:.5f} | c = {titik_tengah_c:.5f} | f(c) = {nilai_fc:.5f}")

        if abs(nilai_fc) < batas_galat:
            break
        elif persamaan_non_linear(batas_bawah) * nilai_fc < 0:
            batas_atas = titik_tengah_c
        else:
            batas_bawah = titik_tengah_c

        iterasi_ke += 1

    return (batas_bawah + batas_atas) / 2.0

akar_persamaan = metode_bagi_dua(2.0, 3.0)
print(f"\n=== HASIL HAMPIRAN AKAR PERSAMAAN: x = {akar_persamaan:.5f} ===")
```
